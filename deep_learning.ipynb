```python
from glob import glob
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from keras.src.layers import Rescaling
from tensorflow.keras.layers import Dense, Dropout, Flatten, Conv2D
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from keras import Input
from keras.src.optimizers import Adam
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
from keras.src.metrics.metrics_utils import confusion_matrix
from sklearn.metrics import precision_score, recall_score


fashion_path_train = glob("../Fashion_MNIST/fashion-mnist_train.csv")
fashion_path_test = glob("../Fashion_MNIST/fashion-mnist_test.csv")
# fish_path
fashion_data_train_ = pd.read_csv(fashion_path_train[0])
fashion_data_test_ = pd.read_csv(fashion_path_test[0])

# fashion_data_train_.shape
# fashion_data_test_.shape

# fashion_data_train_.isnull().sum().sum()
# fashion_data_train_.isna().sum().sum()
# fashion_data_test_.isnull().sum().sum()
# fashion_data_test_.isna().sum().sum()
```

### Preprocessing

```python
# from sklearn.preprocessing import StandardScaler

X_train = fashion_data_train_.drop(['label'], axis = 1)
X_test = fashion_data_test_.drop(['label'], axis = 1)
y_train = fashion_data_train_["label"]
y_test = fashion_data_test_["label"]

y_train_cat = keras.utils.to_categorical(y_train, num_classes=10)
y_test_cat  = keras.utils.to_categorical(y_test,  num_classes=10)
```

### Model Dense, Compile, Train, Predict

```python
model = Sequential([
    Input(shape=(784,)),
    Rescaling(1./255),
    Dense(64, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])


model.compile(optimizer=Adam(learning_rate=0.001), loss='categorical_crossentropy')


model.fit(X_train, y_train_cat, epochs=30, batch_size=32, validation_split=0.1)

model.evaluate(X_test, y_test_cat)

y_predi = model.predict(X_test)

y_pred = np.argmax(y_predi, axis=1)

y_pred
```

```python
print(y_pred)
print(y_test)
```

```python
confusion_m = confusion_matrix(y_test, y_pred, num_classes=10)

precision = precision_score(y_test, y_pred, average="macro")
recall    = recall_score(y_test, y_pred, average="macro")

print("Precision :", precision)
print("Recall    :", recall)


print(f" CONFUSION MATRIX : \n {confusion_m}")
print(f" \n CLASSIFICATION REPORT : \n {classification_report(y_test, y_pred)}")
```

## Test de plusieurs paramètres

### Model A

```python
epochs = [30, 40, 50, 60, 70]

neurons = [32, 64, 128, 256, 512]
```

```python
precisions_A = {}
all_precisions_A = {}

for neuron in neurons:
    for epoch in epochs:
        model_A = Sequential([
            Input(shape=(784,)),
            Rescaling(1./255),
            Dense(neuron, activation="relu"),
            Dense(neuron, activation="relu"),
            Dense(10, activation="softmax")
        ])

        model_A.compile(optimizer=Adam(learning_rate=0.001), loss='categorical_crossentropy')

        model_A.fit(X_train, y_train_cat, epochs=epoch, batch_size=32, validation_split=0.1)

        model_A.evaluate(X_test, y_test_cat)

        y_predi_A = model_A.predict(X_test)

        y_pred_A = np.argmax(y_predi_A, axis=1)

        precisions_A[epoch] = precision_score(y_test, y_pred_A, average="macro")

    all_precisions_A[neuron] = precisions_A
    precisions_A = {}
```

```python
all_precisions_A_df = pd.DataFrame(all_precisions_A)
all_precisions_A_df
```

##### POur ce model les meilleurs paramètres sont : nbre de neurones = 256 et epochs = 40. La précision est de 0.898835
